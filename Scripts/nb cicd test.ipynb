```python
def get_env():
    notebook_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
    components = notebook_path.split('/')
    env = components[components.index('Scripts') -1]
    return env
```

```python
def check_catalog(catalog_name):
    return spark.sql(f"SHOW CATALOGS LIKE '{catalog_name}'").count() > 0
```

```python
# catalog mapping
catalog_mapping = {
    'dev': 'dev_catalog',
    'prod': 'prod_catalog'
}
def get_catalog():
    env = get_env()
    catalog = catalog_mapping[env]
    return catalog
```
